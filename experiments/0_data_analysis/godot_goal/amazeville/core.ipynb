### **B. Data**

```python
from offbench.data.episodes_db import GodotGoalEpisodesDB
from offbench.utils.paths import OFFBENCH_PATH
from typing import Dict

import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
import seaborn as sns
import torch
```

```python
seed = 0

n_episodes = [5, 25, 50, 100]

assets_dir_path = os.path.join(OFFBENCH_PATH, "assets", "environments", "godot_goal", "amazeville")

tasks = [
    "AH - OOO",
    "AH - OOX",
    "AH - XOO",
    "AH - XOX",
    "AL - OOO",
    "AL - OOX",
    "AL - XOO",
    "AL - XOX",
]

task_to_env = {
    "AH - OOO": "amazeville-maze_1-high",
    "AH - OOX": "amazeville-maze_2-high",
    "AH - XOO": "amazeville-maze_3-high",
    "AH - XOX": "amazeville-maze_4-high",
    "AL - OOO": "amazeville-maze_1-low",
    "AL - OOX": "amazeville-maze_2-low",
    "AL - XOO": "amazeville-maze_3-low",
    "AL - XOX": "amazeville-maze_4-low",
}

task_to_background_paths = {
    "AH - OOO": os.path.join(assets_dir_path, "amazeville_high_1_up.png"),
    "AH - OOX": os.path.join(assets_dir_path, "amazeville_high_2_up.png"),
    "AH - XOO": os.path.join(assets_dir_path, "amazeville_high_3_up.png"),
    "AH - XOX": os.path.join(assets_dir_path, "amazeville_high_4_up.png"),
    "AL - OOO": os.path.join(assets_dir_path, "amazeville_low_1_up.png"),
    "AL - OOX": os.path.join(assets_dir_path, "amazeville_low_2_up.png"),
    "AL - XOO": os.path.join(assets_dir_path, "amazeville_low_3_up.png"),
    "AL - XOX": os.path.join(assets_dir_path, "amazeville_low_4_up.png"),
}

task_to_dataset = {
    "AH - OOO": GodotGoalEpisodesDB(task_to_env["AH - OOO"]),
    "AH - OOX": GodotGoalEpisodesDB(task_to_env["AH - OOX"]),
    "AH - XOO": GodotGoalEpisodesDB(task_to_env["AH - XOO"]),
    "AH - XOX": GodotGoalEpisodesDB(task_to_env["AH - XOX"]),
    "AL - OOO": GodotGoalEpisodesDB(task_to_env["AL - OOO"]),
    "AL - OOX": GodotGoalEpisodesDB(task_to_env["AL - OOX"]),
    "AL - XOO": GodotGoalEpisodesDB(task_to_env["AL - XOO"]),
    "AL - XOX": GodotGoalEpisodesDB(task_to_env["AL - XOX"]),
}
```

### **C. Trajectories Visualization**

```python
# to color the trajectories according to the door they go through
door_checkpoints = [(-10.0,0.0),(0.0,0.0),(10.0,0.0)]
colors = ["gold","palegreen","deepskyblue","lavender"]

# to adjust the plot limits depending on the number of episodes
n_episodes_to_adjustments = {
    5: [1.15, 1.125, 1.25, 1.15],
    25: [1.15, 1.125, 1.25, 1.15],
    50: [1.15, 1.125, 1.25, 1.15],
    100: [1.15, 1.125, 1.25, 1.15],
}

for n_episode in n_episodes:

    # it will store the trajectories for each task and the corresponding colors
    task_to_trajectories:Dict[str,Dict[str,torch.Tensor]] = {}
    task_to_trajectories_colors:Dict[str,Dict[str,int]] = {}

    for task,dataset in task_to_dataset.items():
        
        # we just keep the first n_episode trajectories
        ids = dataset.get_ids()
        np.random.seed(seed)
        np.random.shuffle(ids)
        ids = ids[:n_episode]

        # we get the trajectories
        trajectories = {idx: dataset[idx].to_dict()["observation/pos"][0,:,[0,2]] for idx in ids}
        
        # and the corresponding colors
        trajectories_colors = {}
        for idx, positions in trajectories.items():
            positions = [tuple(pos) for pos in positions]
            checkpoint = door_checkpoints[0]
            dist = np.linalg.norm(np.array(positions[0])-np.array(checkpoint))
            for i,pos in enumerate(positions):
                if (i % 2 == 1) and (i < len(positions)-1): continue
                for new_checkpoint in door_checkpoints:
                    new_dist = np.linalg.norm(np.array(pos)-np.array(new_checkpoint))
                    if new_dist < dist:
                        checkpoint = new_checkpoint
                        dist = new_dist
            trajectories_colors[idx] = door_checkpoints.index(checkpoint)
            trajectories_colors[idx] = door_checkpoints.index(checkpoint)
        
        # we store the trajectories and the corresponding colors for the task
        task_to_trajectories[task] = trajectories
        task_to_trajectories_colors[task] = trajectories_colors

    # we set the figure
    fig, axs = plt.subplots(2, 4, figsize=(18,10))
    print("\nN Episodes = {} / Seed = {}".format(n_episode,seed))

    # set the bounds of the plot
    x_min:float = np.min([torch.min(-xy[:,0]).item() for _trajectories in task_to_trajectories.values() for _,xy in _trajectories.items()]) * n_episodes_to_adjustments[n_episode][0]
    x_max:float = np.max([torch.max(-xy[:,0]).item() for _trajectories in task_to_trajectories.values() for _,xy in _trajectories.items()]) * n_episodes_to_adjustments[n_episode][1]
    y_min:float = np.min([torch.min(-xy[:,1]).item() for _trajectories in task_to_trajectories.values() for _,xy in _trajectories.items()]) * n_episodes_to_adjustments[n_episode][2]
    y_max:float = np.max([torch.max(-xy[:,1]).item() for _trajectories in task_to_trajectories.values() for _,xy in _trajectories.items()]) * n_episodes_to_adjustments[n_episode][3]

    for i,task in enumerate(tasks):

        # get the trajectories
        trajectories = task_to_trajectories[task]
        trajectories_colors = task_to_trajectories_colors[task]

        # remove spines
        axs[i//4][i%4].spines['top'].set_visible(False)
        axs[i//4][i%4].spines['right'].set_visible(False)
        axs[i//4][i%4].spines['bottom'].set_visible(False)
        axs[i//4][i%4].spines['left'].set_visible(False)

        # plot trajectories in the corresponding subplot
        axs[i//4][i%4].set_title(task)

        # set the background image
        background = plt.imread(task_to_background_paths[task])
        axs[i//4][i%4].imshow(background)

        # rescale the trajectories
        x_scale = background.shape[0] / (x_max - x_min)
        y_scale = background.shape[1] / (y_max - y_min)

        for idx in trajectories:

            xy = - trajectories[idx]

            x_rescaled = (xy[:,0] - x_min) * x_scale
            y_rescaled = (xy[:,1] - y_min) * y_scale

            color = colors[trajectories_colors[idx]]

            # plot the trajectory
            axs[i//4][i%4].plot(x_rescaled, y_rescaled, color = color, linewidth = 0.5, alpha=0.75)

            # plot circle at the beginning
            axs[i//4][i%4].plot(x_rescaled[0], y_rescaled[0], 'o', color = color, markersize=5, alpha=0.5)

            # plot cross at the end
            axs[i//4][i%4].plot(x_rescaled[-1], y_rescaled[-1], 'x', color = color, markersize=9, alpha=0.5)

        # hide xticks and yticks
        axs[i//4][i%4].set_xticks([])
        axs[i//4][i%4].set_yticks([])

    # space between subplots
    plt.subplots_adjust(wspace=0.1,hspace=0)
    plt.tight_layout()

    # save the plot
    plt.savefig("./plots/trajectories_N{}_S{}.png".format(n_episode,seed),transparent=True)
    plt.savefig("./plots/trajectories_N{}_S{}_alt.png".format(n_episode,seed),transparent=False)

    # show the plot
    plt.show()
```

```python
# to adjust the plot limits depending on the number of episodes
n_episodes_to_adjustments = {
    5: [1.15, 1.125, 1.25, 1.15],
    25: [1.15, 1.125, 1.25, 1.15],
    50: [1.15, 1.125, 1.25, 1.15],
    100: [1.15, 1.125, 1.25, 1.15],
}

for n_episode in n_episodes:

    # it will store the positions for each task
    task_to_positions:Dict[str,torch.Tensor] = {}

    for task,dataset in task_to_dataset.items():
        
        # we just keep the first n_episode trajectories
        ids = dataset.get_ids()
        np.random.seed(seed)
        np.random.shuffle(ids)
        ids = ids[:n_episode]

        # we get the positions
        positions = torch.cat([dataset[idx].to_dict()["observation/pos"][0,:,[0,2]] for idx in ids],dim=0)

        # we store the positions for the task
        task_to_positions[task] = positions

    # we set the figure
    fig, axs = plt.subplots(2, 4, figsize=(18,10))
    print("\n N Episodes = {} / Seed = {}".format(n_episode,seed))
    
    # Set the bounds of the plot
    x_min:float = np.min([torch.min(-xy[:,0]).item() for xy in task_to_positions.values()]) * n_episodes_to_adjustments[n_episode][0]
    x_max:float = np.max([torch.max(-xy[:,0]).item() for xy in task_to_positions.values()]) * n_episodes_to_adjustments[n_episode][1]
    y_min:float = np.min([torch.min(-xy[:,1]).item() for xy in task_to_positions.values()]) * n_episodes_to_adjustments[n_episode][2]
    y_max:float = np.max([torch.max(-xy[:,1]).item() for xy in task_to_positions.values()]) * n_episodes_to_adjustments[n_episode][3]

    for i,task in enumerate(tasks):

        # get the positions
        xy = task_to_positions[task]    

        # Remove spines
        axs[i//4][i%4].spines['top'].set_visible(False)
        axs[i//4][i%4].spines['right'].set_visible(False)
        axs[i//4][i%4].spines['bottom'].set_visible(False)
        axs[i//4][i%4].spines['left'].set_visible(False)
        
        # Plot trajectories in the corresponding subplot
        axs[i//4][i%4].set_title(task)

        # Set the background image
        background = plt.imread(task_to_background_paths[task])
        rotated_background = np.rot90(background,k=2)
        axs[i//4][i%4].imshow(rotated_background)

        # Rescale the trajectories
        x_scale = background.shape[1] / (x_max - x_min)
        y_scale = background.shape[0] / (y_max - y_min)

        x_rescaled = (xy[:,0] - x_min) * x_scale
        y_rescaled = (xy[:,1] - y_min) * y_scale
            
        data = pd.DataFrame({"x": x_rescaled, "y": y_rescaled})

        # plot the density
        kde = sns.kdeplot(
            data=data, x="x", y="y", 
            fill=True, alpha=0.5,
            ax=axs[i//4][i%4], 
            clip=(50.0, 700.0), 
            thresh=0.001, levels=10
        )
        kde.set(xticklabels=[])
        kde.set(yticklabels=[])
        kde.set(xlabel=None)
        kde.set(ylabel=None)
        kde.invert_yaxis()

        # hide xticks and yticks
        axs[i//4][i%4].set_xticks([])
        
        axs[i//4][i%4].set_yticks([])

    # hpace between subplots
    plt.subplots_adjust(wspace=0.1,hspace=0)
    plt.tight_layout()

    # save the plot
    plt.savefig("./plots/densities_N{}_S{}.png".format(n_episode,seed),transparent=True)
    plt.savefig("./plots/densities_N{}_S{}_alt.png".format(n_episode,seed),transparent=False)

    # show the plot
    plt.show()
```

### **D. Metrics**

```python
metrics = {
    "metric": [],
    "n_episode": [],
    "task": [],
    "value": [],
}

for n_episode in n_episodes:

    # AVG SUCCESS
    avg_success = []
    for task,dataset in task_to_dataset.items():
        successes = []
        ids = dataset.get_ids()
        np.random.seed(seed)
        np.random.shuffle(ids)
        ids = ids[:n_episode]
        for idx in ids: successes.append(dataset[idx].to_dict()["done"][0,-1].float().item())
        metrics["metric"].extend(["avg_success"] * len(successes))
        metrics["n_episode"].extend([n_episode] * len(successes))
        metrics["task"].extend([task] * len(successes))
        metrics["value"].extend(successes)

    # AVG LENGTHS
    avg_lengths = []
    for task,dataset in task_to_dataset.items():
        lengths = []
        ids = dataset.get_ids()
        np.random.seed(seed)
        np.random.shuffle(ids)
        ids = ids[:n_episode]
        for idx in ids: lengths.append(len(dataset[idx].to_dict()["observation/pos"][0]))
        metrics["metric"].extend(["avg_length"] * len(lengths))
        metrics["n_episode"].extend([n_episode] * len(lengths))
        metrics["task"].extend([task] * len(lengths))
        metrics["value"].extend(lengths)

    # AVG VELOCITIES
    for task,dataset in task_to_dataset.items():
        velocities = []
        ids = dataset.get_ids()
        np.random.seed(seed)
        np.random.shuffle(ids)
        ids = ids[:n_episode]
        for idx in ids:
            velocities.append(dataset[idx].to_dict()["observation/obs"][0,:,-3:].mean().item())
        metrics["metric"].extend(["avg_velocity"] * len(velocities))
        metrics["n_episode"].extend([n_episode] * len(velocities))
        metrics["task"].extend([task] * len(velocities))
        metrics["value"].extend(velocities)

    # AVG DISTANCES
    for task,dataset in task_to_dataset.items():
        distances = []
        ids = dataset.get_ids()
        np.random.seed(seed)
        np.random.shuffle(ids)
        ids = ids[:n_episode]
        for idx in ids:
            positions = dataset[idx].to_dict()["observation/pos"][0,:,[0,2]]
            distance = 0
            position = positions[0]
            for new_position in positions[1:]:
                distance += torch.norm(position-new_position).float().item()
                position = new_position
            distances.append(distance)
        metrics["metric"].extend(["avg_distance"] * len(distances))
        metrics["n_episode"].extend([n_episode] * len(distances))
        metrics["task"].extend([task] * len(distances))
        metrics["value"].extend(distances)

# ALL METRICS
df = pd.DataFrame(metrics)

fig, ax = plt.subplots(nrows=1, ncols=4, figsize=(18,5))

# reduce space between subplots
plt.subplots_adjust(wspace=0.05,hspace=0)

metric_titles = ["Success Rate", "Episode Length", "Velocity (m/s)", "Covered Distance (m)"]

metric_to_cmap = {
    "avg_success": "PiYG_r",
    "avg_length": "PiYG_r",
    "avg_velocity": "PiYG",
    "avg_distance": "PiYG_r",
}

for i, metric in enumerate(["avg_success", "avg_length", "avg_velocity", "avg_distance"]):

    df_metric = df[df["metric"]==metric][["n_episode", "task", "value"]]
    df_metric = df_metric.groupby(["n_episode", "task"]).agg(["mean", "std"]).reset_index()
    df_metric.columns = ['_'.join(col).rstrip('_') for col in df_metric.columns.values]

    mean_values = df_metric.pivot(index="task", columns="n_episode", values=f"value_mean")
    std_values = df_metric.pivot(index="task", columns="n_episode", values=f"value_std")

    # need to check annot generation as I get the error : ValueError: Unknown format code 'g' for object of type 'str'
    # annot should be mean_values.round(2).astype(str) + " ± " + std_values.round(2).astype(str)
    annot = mean_values.round(2).astype(str) + " ± " + std_values.round(2).astype(str)

    sns.heatmap(
        mean_values,
        ax=ax[i],
        annot=annot,
        annot_kws={"fontsize":8},
        fmt="",
        cbar=False,
        cmap=metric_to_cmap[metric],
    )

    ax[i].set_xlabel("N Episodes")
    ax[i].set_ylabel("")

    if i == 0:
        # pivot y ticks by 90
        ax[i].set_yticklabels(ax[i].get_yticklabels(), rotation=0)

    if i > 0: ax[i].set_yticks([])

    ax[i].set_title(metric_titles[i])

# Space between subplots
plt.tight_layout()

# Save the plot
plt.savefig("./plots/metrics_S{}.png".format(seed),transparent=True)
plt.savefig("./plots/metrics_alt_S{}.png".format(seed),transparent=False)

# Show the plot
plt.show()
```
